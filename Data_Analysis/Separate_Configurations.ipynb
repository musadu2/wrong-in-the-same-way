```python
import os

# Create the input and output file paths
input_file_path = "dft_li3ps4.xyz"
output_folder = "DFT_Li3PS4_Divided"

# Create the output folder if it doesn't exist
os.makedirs(output_folder, exist_ok=True)

# Initialize variables
file_counter = 1
output_file = None

# Open the input file for reading
with open(input_file_path, 'r') as input_file:
    for line in input_file:
        # Check if the line consists of only a number
        if line.strip().isdigit():
            # Close the previous output file if it's open
            if output_file:
                output_file.close()
            
            # Create a new output file
            output_file_path = os.path.join(output_folder, f"training{file_counter}.xyz")
            output_file = open(output_file_path, 'w')
            file_counter += 1
        
        # Write the line to the current output file
        if output_file:
            output_file.write(line)

# Close the last output file if it's open
if output_file:
    output_file.close()

print("File division complete.")
```
